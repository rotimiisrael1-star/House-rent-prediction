```python
import requests

url = "https://propertypro.ng/property-for-rent/in/oyo/ibadan/?search=Ibadan+%2C+Oyo&auto=Ibadan+%2C+Oyo&type=&bedroom=&min_price=&max_price="

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print(response.status_code)
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "lxml")

print(soup.title.text)
```

```python
properties = soup.find_all("div", class_="single-room-text")

print(len(properties))
```

```python
data = []

for prop in properties:
    
    title = prop.find("h4")
    price = prop.find("h3")
    location = prop.find("span")
    
    title = title.text.strip() if title else None
    price = price.text.strip() if price else None
    location = location.text.strip() if location else None
    
    data.append({
        "Property_Type": title,
        "Price": price,
        "Location": location
    })

print(data[:5])
```

```python

from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

driver = webdriver.Chrome()

url = "https://propertypro.ng/property-for-rent/in/oyo/ibadan"

driver.get(url)

time.sleep(5)

properties = driver.find_elements(By.CLASS_NAME, "single-room-text")

data = []

for prop in properties:
    
    title = prop.find_element(By.TAG_NAME, "h4").text
    price = prop.find_element(By.TAG_NAME, "h3").text
    location = prop.find_element(By.TAG_NAME, "span").text
    
    data.append([title, price, location])

driver.quit()

df = pd.DataFrame(data, columns=["Property_Type", "Price", "Location"])

print(df.head())
```
